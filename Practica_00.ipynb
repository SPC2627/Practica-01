### Algunos conceptos básicos para la primera práctica

Cuatro conceptos que os van a aparecer desde el primer momento.

- print(...): muestra algo por pantalla. Es la forma que tiene el programa de "hablar" con quien lo ejecuta. Por ejemplo, print("Hola") escribe la palabra Hola en la pantalla.

```python
print("Hola")
```

- input(): pide un dato por teclado y lo devuelve como texto. Si el dato que esperáis es un número, hay que convertirlo explícitamente con int(...) (para enteros) o float(...) (para decimales). Por ejemplo, edad = int(input()) lee lo que escribe el usuario y lo guarda como un número entero en la variable edad.

```python
print("Hola, ¿cómo te llamas?")
nombre = input()
print("Te llamas " + nombre)
print(f"Confirmo que te llamas {nombre}")
```

- Variables: son como cajas con nombre donde guardáis un dato para usarlo más tarde. A diferencia de otros lenguajes, en Python no hace falta avisar de antemano qué tipo de dato va a guardar una variable: simplemente escribes x = 5 y ya existe.

```python
nombre = "Perico"
edad = 75
peso = 72.4

print(f"El abuelo {nombre} tiene {edad} años y pesa {peso} kilos.")
```

### Cadenas de texto: Ayuda para los ejercicios

La práctica trabaja bastante con cadenas de texto (string), que no son más que texto entre comillas, como "hola mundo". Para el segundo programa y los dos ejercicios finales os bastará con esto:

El operador "in" comprueba si un texto está contenido dentro de otro, y devuelve True o False. Por ejemplo, "mar" in "amarillo" da True, porque "mar" aparece dentro de "amarillo".

```python
"mar" in "amarillo"
```

El método .count(...): se escribe pegado a una variable de tipo cadena (String), como cadena.count(subcadena), y cuenta cuántas veces aparece subcadena dentro de cadena.

```python
texto = "En un lugar de la mancha de cuyo nombre no quiero acordarme..."
texto.count("a")
```

Los métodos .upper() y .lower(): convierten una cadena a mayúsculas o minúsculas. Por ejemplo, "hola".upper() da "HOLA".

```python
texto = "Mi nombre es Julio" # cadena
mayus = texto.upper()        # pongo todo en mayúscula
# print(mayus)
minus = mayus.lower()        # pongo todo en minúsculas
# print(minus)
```


En Python, este tipo de funciones "pegadas" a un dato con un punto (variable.metodo()) se llaman **métodos**, y son una forma muy habitual de operar sobre cadenas, listas y otros tipos de datos: no os preocupéis si el nombre es nuevo, el uso es sencillo.

```python
# Distintos tipos de variables tienen distintos tipos de métodos (o funciones incrustradas)

# help(int)
# help(float)

num  = -5
num.__abs__()
# print(num.__abs__())

text = "HOLA"
text.count("a")
```
